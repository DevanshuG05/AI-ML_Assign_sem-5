```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Create dataset
data = {
    'Name': ['Amit', 'Rahul', 'Sneha', 'Priya', 'Rohan', 'Neha', 'Karan', 'Anjali'],
    'Age': [20, 21, 19, 22, 20, 21, 23, 19],
    'Marks': [85, 78, 92, 88, 76, 95, 81, 89],
    'Attendance': [90, 85, 95, 88, 80, 98, 87, 93]
}

# Create DataFrame
df = pd.DataFrame(data)

# Display dataset
print("Dataset:")
display(df)

# Display first five records
print("First Five Records:")
display(df.head())

# Statistical summary
print("Statistical Summary:")
display(df.describe())

# Basic calculations
print("Average Marks:", df['Marks'].mean())
print("Maximum Marks:", df['Marks'].max())
print("Minimum Marks:", df['Marks'].min())

# Filter students scoring above 85
print("\nStudents scoring above 85:")
display(df[df['Marks'] > 85])

# -----------------------------
# MATPLOTLIB BAR CHART
# -----------------------------

plt.figure(figsize=(8, 5))

plt.bar(df['Name'], df['Marks'])

plt.xlabel('Student Name')
plt.ylabel('Marks')
plt.title('Student Marks')

plt.xticks(rotation=45)
plt.show()


# -----------------------------
# MATPLOTLIB LINE GRAPH
# -----------------------------

plt.figure(figsize=(8, 5))

plt.plot(df['Name'], df['Marks'], marker='o')

plt.xlabel('Student Name')
plt.ylabel('Marks')
plt.title('Marks of Students')

plt.xticks(rotation=45)
plt.show()


# -----------------------------
# SEABORN SCATTER PLOT
# -----------------------------

plt.figure(figsize=(8, 5))

sns.scatterplot(
    x='Attendance',
    y='Marks',
    data=df
)

plt.xlabel('Attendance (%)')
plt.ylabel('Marks')
plt.title('Attendance vs Marks')

plt.show()


# -----------------------------
# SEABORN HEATMAP
# -----------------------------

plt.figure(figsize=(6, 4))

correlation = df[['Age', 'Marks', 'Attendance']].corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm'
)

plt.title('Correlation Heatmap')

plt.show()
```
